```python

glycan_dir = data_dir + '/Glycan OMEtif files/'
peptide_dir = data_dir + '/Peptide OMEtif files/'
metab_dir = data_dir + '/Metabolite OMEtif files/'


glycan_files = os.listdir(glycan_dir)
peptide_files = os.listdir(peptide_dir)
metab_files = os.listdir(metab_dir)
```

Extract glycan intensity values

```python
arg_input = '1*0'

img = int(arg_input.split('*')[0])
channel = int(arg_input.split('*')[1])
print(f'Processing image {img} and channel {channel}')

single_image = metab_files[img]


image = tiff.imread(metab_dir + single_image)
print(f'Image shape: {image.shape}')

### for each pixel in the image, find the intensity of the pixel in the msi image
### create a dataframe for it

output_normal = out_dir + 'metab_msi/'
os.makedirs(output_normal, exist_ok=True)

folder_name = 'metab_' + single_image.split(' ')[0]

print(folder_name)

output_maldi_metab_intensities = output_normal + folder_name + '/'
os.makedirs(output_maldi_metab_intensities, exist_ok=True)

df_maldi_metab = pd.DataFrame(columns=['x', 'y', 'intensity'])

#for channel in range(0, image.shape[0]):
img = image[channel, :, :]
print(img.shape)

for y in range(img.shape[0]):
    for x in range(img.shape[1]):
        intensity = img[y, x]
        df_maldi_metab = pd.concat([df_maldi_metab, pd.DataFrame({'x': [y], 'y': [x], 'intensity': [intensity]})])

df_maldi_metab = df_maldi_metab[df_maldi_metab['intensity'] > 0]
df_maldi_metab['y'] = df_maldi_metab['y'] + 50
df_maldi_metab['x'] = df_maldi_metab['x'] + 50

df_maldi_metab.to_parquet(output_maldi_metab_intensities+f'maldi_metab_channel_{channel}.parquet', index=False)
```
