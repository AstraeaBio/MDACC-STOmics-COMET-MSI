```python
from valis import registration
import os
```

```python
home_dir = os.path.expanduser("~")
data_dir = os.path.join(home_dir, "ext_hd_sammy", "data")
project_dir = os.path.join(home_dir, "ext_hd_sammy", "projects")

out = os.path.join(project_dir, 'out')
slide_src_dir = os.path.join(out, 'matched_images_stomics_COMET')
results_dst_dir = os.path.join(out, "registration_dst_stomics_COMET")
os.makedirs(results_dst_dir, exist_ok=True)


for dir in os.listdir(slide_src_dir):
    ref_img = f'ssDNA_{dir.split("_")[1]}_regist.tif'
    sample_src_dir = os.path.join(slide_src_dir, dir)

    sample_results_dst_dir = os.path.join(results_dst_dir, dir)
    os.makedirs(sample_results_dst_dir, exist_ok=True)
    registered_slide_dst_dir = os.path.join(sample_results_dst_dir, 'registered_slides')
    os.makedirs(registered_slide_dst_dir, exist_ok=True)


    # Create a Valis object and use it to register the slides in slide_src_dir
    registrar = registration.Valis(sample_src_dir, sample_results_dst_dir, align_to_reference=True, reference_img_f=ref_img)
    rigid_registrar, non_rigid_registrar, error_df = registrar.register()
    registrar.warp_and_save_slides(registered_slide_dst_dir)
```
